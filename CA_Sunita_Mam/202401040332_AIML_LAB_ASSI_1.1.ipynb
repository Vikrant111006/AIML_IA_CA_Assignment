Vacuum Cleaner Agent Using Artificial Intelligence

**Aim:**  

To design and implement a Smart Vacuum Cleaner using a Simple Reflex Agent that automatically detects the current condition of an airport environment and performs appropriate cleaning, movement, obstacle avoidance, edge detection, and battery management actions.

---

**Experimental Objective:**

* To understand the working principle of a Simple Reflex Agent and its condition-action behavior.
* To model an airport environment consisting of multiple areas with different dirt levels and floor types.
* To implement condition-based cleaning using different suction modes such as Eco, Normal, and Strong.
* To enable the vacuum cleaner to detect and avoid obstacles present in the airport environment.
* To implement edge detection and edge-cleaning functionality.
* To detect situations where the vacuum becomes stuck and automatically recover using reverse and turning movements.
* To implement battery monitoring and automatic return to the charging station when the battery becomes low.
* To evaluate the performance of the vacuum cleaner using parameters such as dirt removed, cleaning operations, movements, obstacles avoided, battery usage, and performance score.

---

**Expected Outcome:**

* Understand how a Simple Reflex Agent observes the current state of an environment and selects an appropriate action.
* Represent a real-world airport environment using Python data structures.
* Implement condition-based decision-making for an autonomous cleaning agent.
* Understand how sensor information such as dirt level, floor type, obstacles, edges, and battery level affects the agent's actions.
* Implement basic autonomous navigation and recovery mechanisms.
* Measure the performance of an intelligent agent using suitable performance parameters.

---

**Concept:**

An AI agent is a system that observes its environment, makes decisions, and performs actions to achieve a particular goal.

In this experiment:

**Environment** → Airport

1. **Agent →** Smart Vacuum Cleaner
2. **Goal →** Clean dirty airport areas and maintain a clean airport environment
3. **Sensor →** Detect dirt level, obstacles, edges, floor type, and battery level
4. **Action →** Move between airport areas, avoid obstacles, adjust suction power, and clean dirty areas
5. **Decision Rules →** Select the dirtier airport area, choose suitable suction based on dirt level, and return for charging when the battery is low
6. **Performance →** Amount of dirt removed, cleaning efficiency, obstacles avoided, and battery usage

The vacuum continuously senses the airport environment and performs suitable actions until all selected airport areas are clean.

---

**Algorithm:**

**Step 1:** Start the program.

**Step 2:** Create the airport environment with multiple areas.

**Step 3:** Enter the dirt level of each airport area.

**Step 4:** Randomly assign a floor type to each area.

**Step 5:** Randomly generate obstacles in the airport areas.

**Step 6:** Initialize the vacuum battery to 100% and set the charging station.

**Step 7:** Compare the dirt levels of the airport areas and select the area having the highest dirt level.

**Step 8:** Check the current battery level.

**Step 9:** If the battery is 20% or lower, return to the charging station and recharge.

**Step 10:** Check for obstacles. If an obstacle is detected, stop, reverse, and turn.

**Step 11:** Check whether the vacuum is stuck. If it is stuck, reverse and turn to recover.

**Step 12:** Move to the selected airport area if required.

**Step 13:** Sense the current dirt level and floor type.

**Step 14:** Select the appropriate suction mode according to the dirt level.

**Step 15:** Check for an edge and activate edge-cleaning mode if required.

**Step 16:** Perform the cleaning operation and reduce the dirt level.

**Step 17:** Update battery usage, cleaning operations, dirt removed, movements, and performance score.

**Step 18:** Check whether all selected airport areas are clean.

**Step 19:** If the areas are not clean, repeat the process.

**Step 20:** If all areas are clean, display the final airport environment and performance report.

**Step 21:** Stop the program.

**Program:**

```python
import random
import time


# SMART VACUUM CLEANER FOR AN AIRPORT
# USING A SIMPLE REFLEX AGENT

# Airport areas
areas = {
    "A": {
        "name": "Departure Area",
        "dirt": 0,
        "floor": "",
        "obstacle": False,
        "edge": False
    },
    "B": {
        "name": "Arrival Area",
        "dirt": 0,
        "floor": "",
        "obstacle": False,
        "edge": False
    }
}

# Initial agent state
location = "A"
charging_station = "A"
battery = 100
direction = "FORWARD"

# Performance parameters
steps = 0
cleaning_operations = 0
obstacles_avoided = 0
dirt_removed = 0
movement_count = 0
battery_used = 0
edge_cleaning_operations = 0
stuck_events = 0
performance_score = 0



# GET DIRT LEVEL

def get_dirt_level(area):
    while True:
        try:
            dirt = int(
                input(
                    f"Enter dirt level for {areas[area]['name']} (0-100): "
                )
            )

            if 0 <= dirt <= 100:
                return dirt

            print("Enter a value between 0 and 100.")

        except ValueError:
            print("Enter a valid number.")



# DISPLAY AIRPORT ENVIRONMENT

def display_environment():
    print("\n" + "=" * 75)
    print(" " * 25 + "AIRPORT ENVIRONMENT")
    print("=" * 75)

    for area in areas:
        if areas[area]["dirt"] == 0:
            status = "CLEAN"
        else:
            status = "DIRTY"

        obstacle = "YES" if areas[area]["obstacle"] else "NO"
        edge = "YES" if areas[area]["edge"] else "NO"

        print(
            f"{areas[area]['name']:<18} | "
            f"Dirt: {areas[area]['dirt']:>3}% | "
            f"Floor: {areas[area]['floor']:<10} | "
            f"Obstacle: {obstacle:<3} | "
            f"Edge: {edge:<3} | "
            f"{status}"
        )

    print("=" * 75)



# SELECT AREA WITH MORE DIRT

def select_dirtier_area():
    dirt_a = areas["A"]["dirt"]
    dirt_b = areas["B"]["dirt"]

    if dirt_a > dirt_b:
        selected_area = "A"

    elif dirt_b > dirt_a:
        selected_area = "B"

    else:
        selected_area = location

    print(
        f"{areas['A']['name']}: {dirt_a}% dirt | "
        f"{areas['B']['name']}: {dirt_b}% dirt"
    )

    print(
        f"Decision → Select {areas[selected_area]['name']}"
    )

    return selected_area



# SELECT SUCTION MODE

def select_suction_mode(dirt, floor):

    if dirt > 50:

        if battery > 40:
            mode = "STRONG"
            cleaning_power = 35
            battery_cost = 7

        else:
            mode = "NORMAL"
            cleaning_power = 20
            battery_cost = 4

    elif dirt > 20:

        mode = "NORMAL"
        cleaning_power = 20
        battery_cost = 4

    else:

        mode = "ECO"
        cleaning_power = 10
        battery_cost = 2

    # Adjust suction for carpet
    if floor == "CARPET":

        if mode == "ECO" and dirt > 5:
            mode = "NORMAL"
            cleaning_power = 15
            battery_cost = 3

        elif mode == "NORMAL" and battery > 40:
            mode = "STRONG"
            cleaning_power = 25
            battery_cost = 6

    return mode, cleaning_power, battery_cost



# CHECK OBSTACLE

def check_obstacle():

    if areas[location]["obstacle"]:
        print("Obstacle detected!")
        return True

    return False



# AVOID OBSTACLE

def avoid_obstacle():

    global battery
    global battery_used
    global obstacles_avoided
    global direction
    global performance_score

    print("Decision → STOP → REVERSE → TURN")

    # Reverse
    direction = "REVERSE"

    battery -= 1
    battery_used += 1

    # Turn
    direction = random.choice(["LEFT", "RIGHT"])

    battery -= 1
    battery_used += 1

    # Remove obstacle after avoiding it
    areas[location]["obstacle"] = False

    obstacles_avoided += 1
    performance_score += 2

    print(f"New direction → {direction}")
    print("Obstacle avoided.")



# CHECK EDGE

def check_edge():

    global battery
    global battery_used

    edge_detected = random.random() < 0.20

    if edge_detected:

        areas[location]["edge"] = True

        battery -= 1
        battery_used += 1

        print(
            "Edge detected → Edge cleaning activated."
        )

        return True

    else:

        areas[location]["edge"] = False

        return False



# CHECK WHETHER VACUUM IS STUCK

def check_stuck():

    global stuck_events
    global battery
    global battery_used
    global direction
    global performance_score

    stuck = random.random() < 0.10

    if stuck:

        print(
            "Vacuum stuck → Reversing and turning."
        )

        direction = "REVERSE"

        battery -= 2
        battery_used += 2

        direction = random.choice(
            ["LEFT", "RIGHT"]
        )

        stuck_events += 1

        performance_score -= 2

        print(
            f"Recovered → Direction: {direction}"
        )

        return True

    return False



# CLEAN CURRENT AIRPORT AREA

def vacuum_area():

    global battery
    global cleaning_operations
    global dirt_removed
    global battery_used
    global performance_score
    global edge_cleaning_operations

    current_dirt = areas[location]["dirt"]
    floor = areas[location]["floor"]

    mode, cleaning_power, battery_cost = (
        select_suction_mode(
            current_dirt,
            floor
        )
    )

    # Edge cleaning
    if areas[location]["edge"]:

        cleaning_power = min(
            cleaning_power,
            15
        )

        battery_cost = 2

        edge_cleaning_operations += 1

        action = "EDGE CLEANING"

    else:

        action = "NORMAL VACUUMING"

    old_dirt = areas[location]["dirt"]

    areas[location]["dirt"] = max(
        0,
        old_dirt - cleaning_power
    )

    removed = (
        old_dirt -
        areas[location]["dirt"]
    )

    # Update battery
    battery -= battery_cost
    battery_used += battery_cost

    # Update performance
    dirt_removed += removed
    cleaning_operations += 1

    performance_score += removed * 2

    print(f"Action → {action}")

    print(
        f"Suction: {mode} | "
        f"Removed: {removed}% | "
        f"Remaining: {areas[location]['dirt']}% | "
        f"Battery: {battery}%"
    )



# MOVE TO ANOTHER AIRPORT AREA

def move_to_area(target_area):

    global location
    global battery
    global movement_count
    global battery_used
    global performance_score
    global direction

    if location == target_area:
        return

    print(
        f"Decision → Move from "
        f"{areas[location]['name']} to "
        f"{areas[target_area]['name']}"
    )

    direction = "FORWARD"

    location = target_area

    battery -= 3
    battery_used += 3

    movement_count += 1

    performance_score -= 1

    print(
        f"Arrived at {areas[location]['name']} | "
        f"Battery: {battery}%"
    )



# RETURN TO CHARGING STATION

def charge():

    global battery
    global location
    global direction

    print("\nBATTERY LOW")
    print(f"Current battery: {battery}%")

    if location != charging_station:

        print(
            "Returning to charging station..."
        )

        direction = "FORWARD"

        location = charging_station

        print(
            f"Reached {areas[charging_station]['name']}."
        )

    print("Charging...")

    while battery < 100:

        battery += 10

        if battery > 100:
            battery = 100

        print(f"Battery: {battery}%")

        time.sleep(0.1)

    print("Battery fully charged.")



# CHECK WHETHER ALL AIRPORT AREAS ARE CLEAN

def all_areas_clean():

    return (
        areas["A"]["dirt"] == 0
        and
        areas["B"]["dirt"] == 0
    )



# SIMPLE REFLEX AGENT

def simple_reflex_agent():

    global steps

    while True:

        steps += 1

        print("\n" + "=" * 75)
        print(f"STEP {steps}")
        print("=" * 75)

        print(
            f"Location: {areas[location]['name']} | "
            f"Battery: {battery}% | "
            f"Direction: {direction}"
        )

        # Battery check
        if battery <= 20:
            charge()
            continue

        # Check whether cleaning is complete
        if all_areas_clean():

            print(
                "\nALL AIRPORT AREAS ARE CLEAN!"
            )

            break

        # Check obstacle
        if check_obstacle():

            avoid_obstacle()

            continue

        # Check stuck condition
        if check_stuck():

            continue

        # Select the dirtiest area
        target_area = select_dirtier_area()

        # Move if required
        if location != target_area:

            move_to_area(target_area)

            continue

        # Read sensors
        current_dirt = areas[location]["dirt"]
        current_floor = areas[location]["floor"]

        print("\nSENSORS")

        print(
            f"Dirt: {current_dirt}% | "
            f"Floor: {current_floor}"
        )

        # Edge detection
        check_edge()

        # Clean area
        if current_dirt > 0:

            print(
                f"Decision → "
                f"{areas[location]['name']} is dirty. "
                f"Cleaning..."
            )

            vacuum_area()

        else:

            print(
                "Airport area is already clean."
            )



# PROGRAM START

print("\n" + "=" * 75)
print("              SMART VACUUM CLEANER")
print("                    AIRPORT")
print("              SIMPLE REFLEX AGENT")
print("=" * 75)


print("\nINITIAL SETUP")


# Get dirt levels
areas["A"]["dirt"] = get_dirt_level("A")
areas["B"]["dirt"] = get_dirt_level("B")


# Random floor types
areas["A"]["floor"] = random.choice(
    [
        "HARD FLOOR",
        "CARPET"
    ]
)

areas["B"]["floor"] = random.choice(
    [
        "HARD FLOOR",
        "CARPET"
    ]
)


# Random obstacles
areas["A"]["obstacle"] = (
    random.random() < 0.30
)

areas["B"]["obstacle"] = (
    random.random() < 0.30
)


battery = 100


# Display initial environment
print("\nINITIAL AIRPORT ENVIRONMENT")

display_environment()


print(
    f"Charging Station: "
    f"{areas[charging_station]['name']}"
)

print("Initial Battery: 100%")


# Starting decision
print("\nSTARTING DECISION")

location = select_dirtier_area()


print(
    f"Vacuum starts in "
    f"{areas[location]['name']}."
)


time.sleep(1)



# RUN SIMPLE REFLEX AGENT

simple_reflex_agent()



# FINAL PERFORMANCE REPORT

print("\n" + "=" * 75)
print("                     FINAL REPORT")
print("=" * 75)


print(
    "\nAirport cleaning completed successfully!\n"
)


print("PERFORMANCE")
print("-" * 45)


print(
    f"Total Steps          : {steps}"
)

print(
    f"Cleaning Operations  : {cleaning_operations}"
)

print(
    f"Dirt Removed         : {dirt_removed}%"
)

print(
    f"Area Movements       : {movement_count}"
)

print(
    f"Obstacles Avoided    : {obstacles_avoided}"
)

print(
    f"Stuck Events         : {stuck_events}"
)

print(
    f"Edge Cleaning        : {edge_cleaning_operations}"
)

print(
    f"Battery Used         : {battery_used}%"
)

print(
    f"Final Battery        : {battery}%"
)

print(
    f"Performance Score    : {performance_score}"
)


# Final environment
print("\nFINAL AIRPORT ENVIRONMENT")

display_environment()


print("\n" + "=" * 75)
print("             AIRPORT VACUUM CLEANING COMPLETED")
print("=" * 75)
```
