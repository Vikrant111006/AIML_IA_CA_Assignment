**8. Theory of Algorithms:**

**8.1 Breadth First Search (BFS):**

Breadth First Search is an uninformed search algorithm that explores the graph level by level.

It uses a FIFO Queue to maintain the order of exploration. It first visits all neighbouring nodes before exploring nodes at the next level.

**Working:**
1. Insert the starting node into the queue.
2. Remove the first node from the queue.
3. Check whether it is the goal.
4. Add all unvisited neighbouring nodes to the queue.
5. Repeat until the goal is found or the queue becomes empty.

BFS finds the path with the minimum number of edges in an unweighted graph. However, it does not guarantee the minimum travel distance in a weighted graph.

**8.2 Depth First Search (DFS):**

Depth First Search explores one branch as deeply as possible before backtracking.

It uses a Stack (LIFO) or recursion to maintain the search process.

**Working:**
1. Start from the initial node.
2. Visit an unvisited neighbouring node.
3. Continue exploring deeper nodes.
4. Backtrack when no unvisited neighbour is available.
5. Stop when the goal is found.

DFS may use less memory than BFS, but it does not guarantee the shortest or lowest-cost path.

**8.3 Depth Limited Search (DLS):**

Depth Limited Search is a modified version of DFS in which a maximum search depth is specified.

The algorithm explores nodes only up to the given depth limit.

It prevents the search from going indefinitely deep and is useful when the maximum solution depth is known.

In this experiment, the depth limit is set to 3.

**8.4 Iterative Deepening Search (IDS):**

Iterative Deepening Search repeatedly performs Depth Limited Search while increasing the depth limit.

It begins with depth 0 and gradually increases the limit until the goal is found.

It combines the advantages of BFS and DFS by providing systematic shallow-level exploration while using relatively less memory.

**8.5 Bidirectional Search:**

Bidirectional Search performs two searches simultaneously.

One search starts from the initial state, while the other starts from the goal state.

The algorithm stops when both searches meet at a common node.

It can reduce the number of explored nodes when the goal state is known and reverse search is possible.

**8.6 Greedy Best-First Search:**

Greedy Best-First Search selects the node that appears closest to the goal according to a heuristic function.

Its evaluation function is:

**f(n) = h(n)**

Where:
- h(n) = Estimated distance from the current node to the goal.

It can reach a goal quickly, but it ignores the actual distance already travelled and does not guarantee an optimal route.

**8.7 A\* Search:**

A* Search combines the actual cost travelled and the estimated remaining cost.

Its evaluation function is:

**f(n) = g(n) + h(n)**

Where:
- g(n) = Actual cost from the starting node to the current node.
- h(n) = Estimated remaining cost to the goal.

A* is suitable for cost-aware route planning when an admissible heuristic is used.

**8.8 Hill Climbing:**

Hill Climbing is a local-search algorithm that continuously improves the current solution.

It examines neighbouring solutions and moves towards a solution having a lower cost.

It stops when no better neighbouring solution is available.

Its main limitation is that it may become trapped in a local optimum.

**8.9 Genetic Algorithm:**

Genetic Algorithm is an evolutionary optimization technique inspired by natural selection.

It uses a population of candidate solutions and improves them through:

1. Initialization
2. Fitness Evaluation
3. Selection
4. Crossover
5. Mutation
6. Replacement and Evolution

It is useful for optimizing the visiting order of multiple charging stations.

**8.10 Constraint Satisfaction Problem (CSP):**

A Constraint Satisfaction Problem consists of variables, domains and constraints.

In this system:

- Each EV is a variable.
- Each possible station-time combination is a domain value.
- The constraint prevents two EVs from occupying the same station during the same time slot.

Backtracking assigns values to variables and checks whether the constraints are satisfied.

If a conflict occurs, the algorithm reverses the assignment and tries another value.

**9. Python Implementation:**



```python

import math
import heapq
import random
import time
import matplotlib.pyplot as plt
import networkx as nx
from collections import deque

# ============================================================
# SMART EV CHARGING ROUTE AND SLOT ALLOCATION SYSTEM
# ============================================================

START = "Home"
GOAL = "Destination"

# -------------------- ROAD NETWORK ---------------------------

GRAPH = {
    "Home": {"Station A": 2.5, "Station D": 5.0},
    "Station A": {"Home": 2.5, "Station B": 3.0, "Station D": 3.0},
    "Station B": {"Station A": 3.0, "Station C": 3.0,
                  "Station D": 4.0, "Station E": 4.0,
                  "Destination": 8.0},
    "Station C": {"Station B": 3.0, "Station F": 3.5},
    "Station D": {"Home": 5.0, "Station A": 3.0,
                  "Station B": 4.0, "Station E": 2.0,
                  "Station G": 2.5},
    "Station E": {"Station B": 4.0, "Station D": 2.0,
                  "Station F": 3.0, "Station G": 3.5,
                  "Destination": 4.0},
    "Station F": {"Station C": 3.5, "Station E": 3.0,
                  "Destination": 3.2},
    "Station G": {"Station D": 2.5, "Station E": 3.5,
                  "Destination": 3.5},
    "Destination": {"Station B": 8.0, "Station E": 4.0,
                    "Station F": 3.2, "Station G": 3.5}
}

COORDINATES = {
    "Home": (0, 0),
    "Station A": (2, 1),
    "Station B": (5, 1),
    "Station C": (8, 1),
    "Station D": (3, 4),
    "Station E": (5, 4),
    "Station F": (8, 4),
    "Station G": (3, 7),
    "Destination": (7, 7)
}

CHARGING_STATIONS = [
    "Station A", "Station B", "Station C",
    "Station D", "Station E", "Station F", "Station G"
]


# -------------------- COMMON FUNCTIONS ----------------------

def path_cost(path):
    if not path:
        return float("inf")
    return round(sum(
        GRAPH[a][b] for a, b in zip(path, path[1:])
    ), 2)


def reconstruct_path(parent, node):
    path = []
    while node is not None:
        path.append(node)
        node = parent[node]
    return path[::-1]


def heuristic(node):
    x1, y1 = COORDINATES[node]
    x2, y2 = COORDINATES[GOAL]
    return 0.8 * math.sqrt(
        (x1-x2)**2 + (y1-y2)**2
    )


# ============================================================
# 1. BREADTH FIRST SEARCH
# ============================================================

def bfs():
    queue = deque([START])
    parent = {START: None}
    visited = {START}
    explored = []

    while queue:
        current = queue.popleft()
        explored.append(current)

        if current == GOAL:
            return reconstruct_path(parent, GOAL), explored

        for neighbour in GRAPH[current]:
            if neighbour not in visited:
                visited.add(neighbour)
                parent[neighbour] = current
                queue.append(neighbour)

    return None, explored


# ============================================================
# 2. DEPTH FIRST SEARCH
# ============================================================

def dfs():
    stack = [START]
    parent = {START: None}
    visited = {START}
    explored = []

    while stack:
        current = stack.pop()
        explored.append(current)

        if current == GOAL:
            return reconstruct_path(parent, GOAL), explored

        for neighbour in reversed(list(GRAPH[current])):
            if neighbour not in visited:
                visited.add(neighbour)
                parent[neighbour] = current
                stack.append(neighbour)

    return None, explored


# ============================================================
# 3. DEPTH LIMITED SEARCH
# ============================================================

def dls(limit):
    explored = []

    def search(node, path):
        explored.append(node)

        if node == GOAL:
            return path

        if len(path) - 1 >= limit:
            return None

        for neighbour in GRAPH[node]:
            if neighbour not in path:
                result = search(
                    neighbour, path + [neighbour]
                )
                if result:
                    return result

        return None

    return search(START, [START]), explored


# ============================================================
# 4. ITERATIVE DEEPENING SEARCH
# ============================================================

def ids(max_depth=10):
    all_explored = []

    for depth in range(max_depth + 1):
        path, explored = dls(depth)
        all_explored.extend(explored)

        if path:
            return path, all_explored

    return None, all_explored


# ============================================================
# 5. BIDIRECTIONAL SEARCH
# ============================================================

def bidirectional_search():
    forward = deque([START])
    backward = deque([GOAL])

    parent_f = {START: None}
    parent_b = {GOAL: None}

    visited_f = {START}
    visited_b = {GOAL}

    explored = []
    meeting = None

    while forward and backward and meeting is None:

        # Forward search
        for _ in range(len(forward)):
            current = forward.popleft()
            explored.append(current)

            if current in visited_b:
                meeting = current
                break

            for neighbour in GRAPH[current]:
                if neighbour not in visited_f:
                    visited_f.add(neighbour)
                    parent_f[neighbour] = current
                    forward.append(neighbour)

                    if neighbour in visited_b:
                        meeting = neighbour
                        break

            if meeting is not None:
                break

        if meeting is not None:
            break

        # Backward search
        for _ in range(len(backward)):
            current = backward.popleft()
            explored.append(current)

            if current in visited_f:
                meeting = current
                break

            for neighbour in GRAPH[current]:
                if neighbour not in visited_b:
                    visited_b.add(neighbour)
                    parent_b[neighbour] = current
                    backward.append(neighbour)

                    if neighbour in visited_f:
                        meeting = neighbour
                        break

            if meeting is not None:
                break

    if meeting is None:
        return None, explored

    left = reconstruct_path(parent_f, meeting)

    right = []
    current = parent_b[meeting]

    while current is not None:
        right.append(current)
        current = parent_b[current]

    return left + right, explored


# ============================================================
# 6. GREEDY BEST-FIRST SEARCH
# ============================================================

def greedy_best_first():
    pq = [(heuristic(START), START)]
    parent = {START: None}
    visited = set()
    explored = []

    while pq:
        _, current = heapq.heappop(pq)

        if current in visited:
            continue

        visited.add(current)
        explored.append(current)

        if current == GOAL:
            return reconstruct_path(parent, GOAL), explored

        for neighbour in GRAPH[current]:
            if neighbour not in visited:
                if neighbour not in parent:
                    parent[neighbour] = current

                heapq.heappush(
                    pq, (heuristic(neighbour), neighbour)
                )

    return None, explored


# ============================================================
# 7. A* SEARCH
# ============================================================

def a_star():
    pq = [(heuristic(START), 0, START)]

    parent = {START: None}
    distance = {START: 0}
    explored = []

    while pq:
        f, g, current = heapq.heappop(pq)

        if g != distance[current]:
            continue

        explored.append(current)

        if current == GOAL:
            return reconstruct_path(parent, GOAL), explored

        for neighbour, weight in GRAPH[current].items():
            new_cost = g + weight

            if new_cost < distance.get(
                neighbour, float("inf")
            ):
                distance[neighbour] = new_cost
                parent[neighbour] = current

                heapq.heappush(
                    pq,
                    (new_cost + heuristic(neighbour),
                     new_cost, neighbour)
                )

    return None, explored


# ============================================================
# DIJKSTRA - HELPER FOR MULTI-STATION OPTIMIZATION
# ============================================================

def dijkstra(start, goal):
    pq = [(0, start)]
    distance = {start: 0}
    parent = {start: None}

    while pq:
        cost, current = heapq.heappop(pq)

        if cost != distance[current]:
            continue

        if current == goal:
            return reconstruct_path(parent, goal), cost

        for neighbour, weight in GRAPH[current].items():
            new_cost = cost + weight

            if new_cost < distance.get(
                neighbour, float("inf")
            ):
                distance[neighbour] = new_cost
                parent[neighbour] = current
                heapq.heappush(
                    pq, (new_cost, neighbour)
                )

    return None, float("inf")


# ============================================================
# MULTI-STATION ROUTE COST
# ============================================================

def tour_cost(route):
    complete_route = [START] + route + [START]
    total = 0

    for a, b in zip(complete_route, complete_route[1:]):
        _, cost = dijkstra(a, b)
        total += cost

    return round(total, 2)


# ============================================================
# 8. HILL CLIMBING
# ============================================================

def hill_climbing():
    current = CHARGING_STATIONS[:]
    current_cost = tour_cost(current)

    while True:
        best_route = current[:]
        best_cost = current_cost

        for i in range(len(current)):
            for j in range(i + 1, len(current)):
                neighbour = current[:]

                neighbour[i], neighbour[j] = (
                    neighbour[j], neighbour[i]
                )

                cost = tour_cost(neighbour)

                if cost < best_cost:
                    best_route = neighbour
                    best_cost = cost

        if best_cost >= current_cost:
            return current, current_cost

        current = best_route
        current_cost = best_cost


# ============================================================
# 9. GENETIC ALGORITHM
# ============================================================

def crossover(p1, p2):
    size = len(p1)
    left, right = sorted(
        random.sample(range(size), 2)
    )

    child = [None] * size
    child[left:right+1] = p1[left:right+1]

    remaining = [x for x in p2 if x not in child]
    index = 0

    for i in range(size):
        if child[i] is None:
            child[i] = remaining[index]
            index += 1

    return child


def genetic_algorithm():
    random.seed(42)

    population = []

    for _ in range(30):
        route = CHARGING_STATIONS[:]
        random.shuffle(route)
        population.append(route)

    best_route = None
    best_cost = float("inf")

    for generation in range(50):

        population.sort(key=tour_cost)

        if tour_cost(population[0]) < best_cost:
            best_route = population[0][:]
            best_cost = tour_cost(best_route)

        new_population = population[:4]

        while len(new_population) < 30:

            p1, p2 = random.sample(population[:15], 2)
            child = crossover(p1, p2)

            if random.random() < 0.15:
                i, j = random.sample(range(7), 2)
                child[i], child[j] = child[j], child[i]

            new_population.append(child)

        population = new_population

    return best_route, best_cost


# ============================================================
# 10. CSP USING BACKTRACKING
# ============================================================

EV_DOMAINS = {
    "EV1": [
        ("Station A", "09:00-10:00"),
        ("Station B", "09:00-10:00"),
        ("Station A", "10:00-11:00")
    ],

    "EV2": [
        ("Station B", "09:00-10:00"),
        ("Station C", "09:00-10:00"),
        ("Station C", "10:00-11:00")
    ],

    "EV3": [
        ("Station A", "10:00-11:00"),
        ("Station C", "10:00-11:00"),
        ("Station C", "11:00-12:00")
    ],

    "EV4": [
        ("Station A", "09:00-10:00"),
        ("Station B", "10:00-11:00"),
        ("Station C", "11:00-12:00")
    ]
}


def csp_backtracking():
    vehicles = list(EV_DOMAINS.keys())
    assignment = {}

    def backtrack(index):
        if index == len(vehicles):
            return assignment.copy()

        vehicle = vehicles[index]

        for slot in EV_DOMAINS[vehicle]:

            if slot not in assignment.values():
                assignment[vehicle] = slot

                result = backtrack(index + 1)

                if result is not None:
                    return result

                del assignment[vehicle]

        return None

    return backtrack(0)


# ============================================================
# GRAPH VISUALIZATION
# ============================================================

def visualize_graph():
    G = nx.Graph()

    for node in GRAPH:
        for neighbour, weight in GRAPH[node].items():
            G.add_edge(node, neighbour, weight=weight)

    pos = {
        "Home": (0, 0),
        "Station A": (2, 1),
        "Station B": (5, 1),
        "Station C": (8, 1),
        "Station D": (3, 4),
        "Station E": (5, 4),
        "Station F": (8, 4),
        "Station G": (3, 7),
        "Destination": (7, 7)
    }

    plt.figure(figsize=(12, 7))

    nx.draw(
        G, pos,
        with_labels=True,
        node_size=2200,
        node_color="lightblue",
        font_size=8,
        font_weight="bold"
    )

    labels = nx.get_edge_attributes(G, "weight")

    nx.draw_networkx_edge_labels(
        G, pos, edge_labels=labels
    )

    plt.title("Smart EV Charging Road Network")
    plt.axis("off")
    plt.show()


# ============================================================
# MAIN PROGRAM
# ============================================================

def main():

    print("=" * 65)
    print("SMART EV CHARGING ROUTE AND SLOT ALLOCATION SYSTEM")
    print("=" * 65)

    algorithms = [
        ("BFS", bfs),
        ("DFS", dfs),
        ("DLS", lambda: dls(3)),
        ("IDS", ids),
        ("Bidirectional Search", bidirectional_search),
        ("Greedy Best-First", greedy_best_first),
        ("A* Search", a_star)
    ]

    print("\n1. SINGLE DESTINATION ROUTE SEARCH")
    print("-" * 65)

    for name, algorithm in algorithms:

        start_time = time.perf_counter()

        path, explored = algorithm()

        elapsed = (
            time.perf_counter() - start_time
        ) * 1000

        print("\nAlgorithm:", name)

        print(
            "Route:",
            " -> ".join(path) if path else "Not Found"
        )

        print(
            "Travel Cost:",
            path_cost(path) if path else "N/A",
            "km"
        )

        print("Expanded Nodes:", len(explored))

        print(
            "Execution Time:",
            round(elapsed, 4), "ms"
        )

    # MULTI-STATION OPTIMIZATION

    print("\n2. MULTI-STATION ROUTE OPTIMIZATION")
    print("-" * 65)

    initial = CHARGING_STATIONS[:]

    print("Initial Route:", " -> ".join(initial))
    print("Initial Tour Cost:", tour_cost(initial), "km")

    hill_route, hill_cost = hill_climbing()

    print("\nHill Climbing:")
    print("Optimized Route:", " -> ".join(hill_route))
    print("Optimized Cost:", hill_cost, "km")

    ga_route, ga_cost = genetic_algorithm()

    print("\nGenetic Algorithm:")
    print("Best Route:", " -> ".join(ga_route))
    print("Best Cost:", ga_cost, "km")

    # CSP

    print("\n3. CHARGING SLOT ALLOCATION USING CSP")
    print("-" * 65)

    allocation = csp_backtracking()

    if allocation:
        for vehicle, (station, slot) in allocation.items():
            print(
                vehicle, "->", station, "|", slot
            )
    else:
        print("No valid allocation found.")

    print("\n4. FINAL SUMMARY")
    print("-" * 65)
    print("Route Finding: BFS, DFS, DLS, IDS, Bidirectional,")
    print("               Greedy Best-First and A*")
    print("Route Optimization: Hill Climbing and Genetic Algorithm")
    print("Slot Allocation: CSP with Backtracking")

    print("\nProgram executed successfully.")


# ============================================================
# EXECUTION
# ============================================================

visualize_graph()
main()
```

**Analysis:**

BFS explores the graph level by level. It finds a route containing the minimum number of edges, but the selected route is not necessarily the minimum-distance route.

**2. Depth First Search:**

DFS explores deeper nodes before backtracking. Its performance depends on the order of neighbouring nodes and it does not guarantee an optimal solution.

**3. Depth Limited Search:**

DLS restricts the search depth to 3. It successfully finds the destination in this experiment.

**4. Iterative Deepening Search:**

IDS repeatedly performs depth-limited searches. It requires more repeated exploration but eventually finds the solution at depth 3.

**5. Bidirectional Search:**

Bidirectional Search explores from both Home and Destination. It reduces the number of explored nodes in this example.

**6. Greedy Best-First Search:**

Greedy Best-First Search uses the estimated distance to the destination. It finds the 11.0 km route through Station D and Station E, but it does not guarantee an optimal solution in general.

**7. A* Search:**

**12. Constraint Satisfaction Problem Analysis:**

The charging-slot allocation problem is modeled as a Constraint Satisfaction Problem.

**A. Variables:**

EV1, EV2, EV3 and EV4.

**B. Domains:**

Each vehicle has a predefined list of available station-time combinations.

**C. Constraint:**

No two electric vehicles can occupy the same charging station during the same time slot.

**D. Backtracking Process:**

1. Select the first electric vehicle.
2. Assign an available charging station and time slot.
3. Check whether the assignment violates any existing constraint.
4. If valid, move to the next vehicle.
5. If invalid, try another available combination.
6. Continue until all vehicles receive valid assignments.

**E. Result:**

The algorithm successfully allocates charging slots to all four vehicles.

No station-time combination is repeated.

**F. Observation:**

CSP is suitable for charging-slot allocation because it handles scheduling requirements and ensures that the defined constraints are satisfied.

**14. Critical Analysis & Evaluation:**

The experiment demonstrates how different Artificial Intelligence search algorithms can be applied to electric vehicle route planning and charging management.

**1. Uninformed Search:**

Uninformed algorithms explore the search space without estimating the remaining distance to the destination.

BFS provides systematic exploration, while DFS explores deeper paths. DLS and IDS control the search depth. Bidirectional Search reduces exploration by searching from both endpoints.

However, these methods do not directly optimize weighted travel distance.

**2. Informed Search:**

Informed algorithms use heuristic information to guide exploration.

Greedy Best-First Search uses estimated remaining distance, whereas A* combines actual travelled cost and estimated remaining cost.

A* is more appropriate for weighted route planning because it considers both factors.

**3. Local Search:**

Hill Climbing improves an existing route by exploring neighbouring solutions.

It is simple to implement but may get trapped in a local optimum.

**4. Genetic Algorithm:**

Genetic Algorithm explores multiple candidate routes simultaneously. Selection, crossover and mutation help generate improved solutions.

It is useful for multi-station route optimization, although its performance depends on population size, mutation rate and number of generations.

**5. Constraint Satisfaction:**

CSP solves the charging appointment allocation problem by assigning valid station-time combinations to EVs.

It prevents scheduling conflicts and is suitable for systems involving multiple vehicles and limited charging resources.

**5. Overall Evaluation:**

The experiment indicates that no single algorithm is sufficient to solve every aspect of smart EV charging management.

A combination of informed search, local optimization, evolutionary algorithms and constraint satisfaction provides a more complete solution.

A* is suitable for route planning, Hill Climbing and Genetic Algorithm support multi-station optimization, and CSP handles charging-slot allocation.
